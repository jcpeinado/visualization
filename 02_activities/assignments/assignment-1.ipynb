```python
#Subplots
x1 = np.arange(50)
y1 = np.random.randint(0,75,50)
x2 = np.array(["Luffy","Zoro","Nami","Usopp","Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
fig, (ax1, ax2) = plt.subplots(
    ncols=2,
    nrows=1,
    figsize=(7,3)
)

ax1.scatter(x1,y1)
ax2.bar(x2,y2)
fig.show()
ax2.grid()
```

```python
# Create a subplot mosaic
fig, someaxes = plt.subplot_mosaic(
    [['ax1', 'ax3'],
     ['ax2', 'ax3']],
    figsize=(7, 4)
)

# Plot different types of graphs
someaxes["ax1"].scatter(x1, y1)  # Scatter plot
someaxes["ax2"].bar(x2, y2)      # Bar chart
someaxes["ax3"].plot(x1, y1)     # Line plot

# Show the plot
plt.show()
```

```python
# Create a subplot mosaic
fig, someaxes = plt.subplot_mosaic(
    [['ax1', 'ax3'],
     ['ax2', 'ax3']],
    figsize=(7, 4),
    layout = "constrained"
)

# Plot different types of graphs
someaxes["ax1"].scatter(x1, y1)  # Scatter plot
someaxes["ax2"].bar(x2, y2)      # Bar chart
someaxes["ax3"].plot(x1, y1)     # Line plot

#Axis Labels
someaxes["ax1"].set_xlabel('A Big Label', fontsize=18)
someaxes["ax2"].set_xlabel('Another Label', fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize=18)

# Show the plot
plt.show()
```

```python

# Sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

#define axis object
fig, ax = plt.subplots(
    figsize=(7,3)
)

ax.bar (x,y1,color='indigo')
ax.plot(x,y2,color='red')
```

```python
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x,y2, color ='red')

ax.errorbar(x,y2,
            yerr=y2_sd,
            fmt="none",
            ecolor = 'indigo',
            elinewidth = 2,
            capsize = 4,
            capthick = 2)
```

```python
#Class_3

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

from PIL import Image
import requests
from io import BytesIO
```

```python
x= np.array(["Luffy","Zoro","Nami","Usopp","Sanji"])
y2= np.array([170,100,90,120,50])

response = requests.get("https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png")
image_file = BytesIO(response.content)
image = Image.open(image_file)
```

```python
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color = "red")

ax_image = fig.add_axes([0.1, 0.11, 0.15, 0.25])

ax_image.imshow(image)
ax_image.axis('off')
```
